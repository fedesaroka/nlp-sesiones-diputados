```python
# 1. Lista exhaustiva de basura descubierta
stopwords_iteracion_1 = [
    # Pronombres/Preposiciones rebeldes
    "él", "es", "de", "con", "pro", "re", "sr", "ca",
    # Basura de OCR y ligaduras
    "ción", "cación", "modi", "rmativa", "apruer", "rrupción", "inte", "pala", "bra", "bue",
    # Jerga Parlamentaria procedimental
    "votar", "resultar", "afirmativa", "negativa", "nominal", "observación", "cuestión", 
    "privilegio", "asuntos", "presidencia", "asentimiento", "expediente", "resolución", 
    "moción", "consideración", "particular", "general", "interrupción", "preferencia", 
    "tabla", "tratamiento", "reglamento", "despacho", "rechazado", "venia", "conceder",
    "abstención", "tablero", "electrónico", "secretaría", "lectura", "orador", "silencio",
    "resultado", "labor", "leer", "ausente", "instante", "formulario", "formulado",
    "aclaración", "homenaje", "permitir", "exposición", "autorización", "práctica",
    # Provincias/Ubicaciones (Opcional, pero NMF las agrupa sin sentido)
    "aires", "ciudad", "autónoma", "capital", "federal", "santa", "córdoba", "jujuy", 
    "mendoza", "ríos", "chaco", "corrientes", "negro", "cruz", "catamarca",
    # Apellidos de legisladores que dominaron NMF y LDA
    "garcía", "gonzález", "gutiérrez", "díaz", "rodríguez", "pérez", "martínez",
    "camaño", "moreau", "lospennato", "ritondo", "heller", "giudici", "ferraro",
    "iglesias", "bregman", "tetaz", "penacca", "lavagna", "plá", "caño"
]

# 2. Arreglo rápido del caracter raro 'ﬁ' en Pandas
print("Limpiando ligaduras tipográficas...")
df['texto_limpio'] = df['texto_limpio'].str.replace('ﬁ', 'fi')
textos = df['texto_limpio'].dropna().tolist()

# 3. Vectorización pasándole las nuevas stopwords
print("Vectorizando para LDA (Count)...")
# Bajamos max_df a 0.85 para ignorar palabras súper frecuentes
tf_vectorizer = CountVectorizer(max_df=0.85, min_df=15, stop_words=stopwords_iteracion_1)
tf_matrix = tf_vectorizer.fit_transform(textos)
tf_feature_names = tf_vectorizer.get_feature_names_out()

print("Vectorizando para NMF (TF-IDF)...")
tfidf_vectorizer = TfidfVectorizer(max_df=0.85, min_df=15, stop_words=stopwords_iteracion_1)
tfidf_matrix = tfidf_vectorizer.fit_transform(textos)
tfidf_feature_names = tfidf_vectorizer.get_feature_names_out()
```

```python
def print_top_words(model, feature_names, n_top_words=10):
    for topic_idx, topic in enumerate(model.components_):
        message = f"Tópico #{topic_idx}: "
        message += " ".join([feature_names[i] for i in topic.argsort()[:-n_top_words - 1:-1]])
        print(message)
    print()

k_topicos = 15

# --- Entrenar LDA ---
print(f"Entrenando LDA con {k_topicos} tópicos...")
lda_model = LatentDirichletAllocation(n_components=k_topicos, random_state=42, n_jobs=-1)
lda_model.fit(tf_matrix)

print("\n--- Resultados LDA ---")
print_top_words(lda_model, tf_feature_names)

# --- Entrenar NMF ---
print(f"\nEntrenando NMF con {k_topicos} tópicos...")
nmf_model = NMF(n_components=k_topicos, random_state=42, init='nndsvda')
nmf_model.fit(tfidf_matrix)

print("\n--- Resultados NMF ---")
print_top_words(nmf_model, tfidf_feature_names)
```

```python
from bertopic import BERTopic
from sentence_transformers import SentenceTransformer
from umap import UMAP
from hdbscan import HDBSCAN
import random

# Tomamos una muestra muy chica para testear que la librería ande
random.seed(42)
textos_prueba = random.sample(textos, 5000)

print("Cargando modelo multilingüe...")
embedding_model = SentenceTransformer("paraphrase-multilingual-MiniLM-L12-v2")
umap_model = UMAP(n_neighbors=15, n_components=5, min_dist=0.0, metric='cosine', random_state=42)
hdbscan_model = HDBSCAN(min_cluster_size=50, metric='euclidean', cluster_selection_method='eom', prediction_data=True)

topic_model = BERTopic(
    embedding_model=embedding_model,
    umap_model=umap_model,
    hdbscan_model=hdbscan_model,
    language="spanish",
    calculate_probabilities=False # Ahorra RAM
)

print("Entrenando prueba rápida de BERTopic...")
topics, probs = topic_model.fit_transform(textos_prueba)
print("¡Funciona perfecto! Tópicos generados:", len(topic_model.get_topic_info()) - 1)
```
